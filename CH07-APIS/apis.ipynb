```python
import requests
import json
```

## **GET**

```python
url = "https://jsonplaceholder.typicode.com/posts/2"

response = requests.get(url)
response        #o/p:<Response [200]> 200 is status code
#status code
print(response.status_code)

#url
print(response.url)

#default headers
print(response.headers)

#response body
print(response.text)
#or
print(response.json())

#parse json response -->for this import json
data=response.json()
print(json.dumps(data, indent=2))
```

## **POST**

```python
url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title":"foo",
    "body":"bar",
    "userId":1   
}

#Method 1: using json parameter
response=requests.post(url, json=data)
print(response.status_code)
print(json.dumps(response.json(),indent=2))

#Method 2: using Form data

form_data = {
    "title":"xyz",
    "body":"abc",
    "userId":1
}

form_data = requests.post(url, data)
print(response.status_code)
print(json.dumps(response.json(),indent=2))
```

## **Query Parameters**

```python
url = "https://jsonplaceholder.typicode.com/posts"

params ={
    "userId": 1,
    "id":1
}

response=requests.get(url, params=params)
print(response.status_code)
print(json.dumps(response.json(),indent=2))


# "https://jsonplaceholder.typicode.com/posts?userId=1&id=1"
```
