```python
from pydantic import BaseModel
```

```python
class employee_schema(BaseModel):
    name:str
    id:int
    department:str
```

```python
obj=employee_schema(name="Sanagmesh",id=22,department="data")
print(obj)
```

```python
obj=employee_schema(name="Sanagmesh",id=22,department="data")
obj
```

```python
obj2=employee_schema(**{"name":"Sanagmesh","id":22,"department":"data"})
obj2
```

## **coercion v/s strict types**

## **corecion**

```python
obj3=employee_schema(**{"name":"Sanagmesh","id":"22","department":"data"})
obj3
```

## **Strict Types**

```python
from pydantic import StrictInt
```

```python
class employee_schema(BaseModel):
    name:str
    id:StrictInt
    department:str
    
obj3=employee_schema(**{"name":"Sanagmesh","id":22,"department":"data"})
obj3                    #This time it throws error when we pass "id":"22"
```
