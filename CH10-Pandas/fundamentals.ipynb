### **Series**
It is used for one dimentional arrays

```python
import pandas as pd
```

Creating Series

```python
employees = pd.Series(["Om","Paddya"])

print(employees)
```

Here 0,1 are indexes and Om, Paddya are actual data

We can access the data using indexes

```python
print(employees[0])
print(employees[1])
```

We can have custome Indexes

```python
students = pd.Series(["Rohan", "Sammya"], index=[101,102])

print(students)
```

Here 101 and 102 are custome indexes.
We can access actual data using those custom labels

```python
print(students[101])
print(students[102])
```

#### **Note: For series we pass lists**

### **DataFrame**

```python
data = {
    "EmployeeID": [101, 102, 103, 104],
    "EmployeeName": ["Rahul", "Priya", "Amit", "Sneha"],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Salary": [50000, 45000, 60000, 55000]
}

df = pd.DataFrame(data)

print(df)
```

Here 0,1,2,3 are indexes which produced by pandas by default

We can have custom indexes here also like Series

```python
df = pd.DataFrame(data, index=["E01","E02","E03","E04"])

print(df)
```

We can access single columns

```python
#DataFrame
print(df[["EmployeeID"]])
```

```python
#Series
print(df["EmployeeID"])
```

##### **Pandas Datatypes**

```python
print(df.dtypes)
```
