### **1. Creating a DataFrame using a Dictionary**

```python
import pandas as pd

data = {
    "EmployeeID": [101, 102, 103],
    "EmployeeName": ["Rahul", "Priya", "Amit"],
    "Department": ["IT", "HR", "IT"],
    "Salary": [50000, 45000, 60000]
}

df = pd.DataFrame(data)

print(df)
```

- Dictionary keys become column names.
- Dictionary values become column data.
- Each position in the lists represents a row.
- Pandas automatically generates indexes if you don't provide them.
Important rule: All lists must have the same length when constructing a DataFrame this way. Otherwise, Pandas raises a ValueError.

### **2. Creating a DataFrame using a List of Lists**

Here we are providing column names explicitly using columns function

```python
data = [
    [101, "Rahul", "IT", 50000],
    [102, "Priya", "HR", 45000],
    [103, "Amit", "IT", 60000]
]

df2=pd.DataFrame(
    data,
    columns=["EmployeeID", "EmployeeName", "Department", "Salary"]
)

print(df2)
```

### **3. Creating a DataFrame using a List of Dictionaries**

```python
data = [
    {"ID":101,"Name":"Om","City":"Belagavi"},
    {"ID":102,"Name":"Paddya","City":"Mysore"},
    {"ID":103,"Name":"Rohan","City":"Bangalore"}
]

#Each dictionary represents one record.

df3 = pd.DataFrame(data)

print(df3)
```

What happens if some dictionaries have missing keys?

```python
data = [
    {"ID":101,"Name":"Om","City":"Belagavi"},
    {"ID":102,"Name":"Paddya"},
    {"ID":103,"Name":"Rohan","City":"Bangalore"}
]

#Here in 2nd dictionary we have missing key city

df4 = pd.DataFrame(data)

print(df4)
```

Pandas automatically inserts NaN where a dictionary doesn't contain a particular key.

### **4. Creating a DataFrame using NumPy Arrays**

NumPy is another Python library primarily used for numerical computing and working with arrays.
Pandas is built on top of NumPy concepts and integrates closely with NumPy arrays.

It's almost like list of lists

```python
import pandas as pd
import numpy as np

data = np.array([
    [101, 50000],
    [102, 45000],
    [103, 60000]
])

print(data)
```

Now convert it into a DataFrame.

```python
df = pd.DataFrame(data, columns=["EmployeeID", "Salary"])
print(df)
```

Here:
- np.array() creates a NumPy array.
- pd.DataFrame() converts that array into a DataFrame.
- columns specifies the column names.

### **5. Creating a DataFrame from Pandas Series**

```python
#We have 3 series
employee_id = pd.Series([101, 102, 103])

employee_name = pd.Series(["Rahul", "Priya", "Amit"])

salary = pd.Series([50000, 45000, 60000])

#We can combine these Series into one DataFrame.
df = pd.DataFrame({
    "EmployeeID": employee_id,
    "EmployeeName": employee_name,
    "Salary": salary
})

print(df)
```

Notice that we used a dictionary again.
But this time, dictionary values are Series instead of ordinary Python lists.

**Important: What happens if Series indexes differ?**

```python
employee_id = pd.Series(
    [101, 102, 103],
    index=[0, 1, 2]
)

employee_name = pd.Series(
    ["Rahul", "Priya", "Amit"],
    index=[0, 2, 3]
)

df = pd.DataFrame({
    "EmployeeID": employee_id,
    "EmployeeName": employee_name
})

print(df)
```

### **7. Adding New Columns to an Existing DataFrame**

```python
df["Department"] = ["IT","IT","Embedded"]
print(df)
```

Adding a column using an existing column\
Suppose we want to calculate an annual bonus equal to 10% of salary.

```python
df["Bonus"]= df["Salary"] * 0.10
print(df)
```

Notice something important:\
We didn't use a loop.\
Pandas automatically performed the multiplication for every row.\
This is called vectorized operation.\
Vectorization is an important concept for performance optimization.

### **8. Adding Rows to a DataFrame**

```python
new_employee = pd.DataFrame({
    "EmployeeID": [104],
    "EmployeeName": ["Sangamesh"],
    "City":["Belavi"],
    "Salary": [500],
    "Department":["IT"],
    "Bonus":[50.0]
})

df = pd.concat(
    [df, new_employee],
    ignore_index=True
)

print(df)
```

- df: Existing DataFrame.
- new_employee: DataFrame containing the new row.
- ignore_index=True: Creates a fresh continuous index.\
Important: The older DataFrame.append() method was deprecated and removed in Pandas 2.0. Use pd.concat() instead.


### **9. Creating DataFrames with Custom Indexes**

We briefly covered custom indexes in the previous lesson. Let's understand the syntax more clearly.\
We can also change the index of an existing DataFrame:

```python
df.index=["Emp1","Emp2","Emp3","Emp4"]

print(df)
```

And reset the index:

```python
df = df.reset_index()

print(df)
```

```python
df.drop(columns=["level_0","index"],inplace=True)
print(df)
```

```python
df = df.reset_index(drop=True)

print(df)
```

Here:
- reset_index() resets the index.
- drop=True prevents the previous index from becoming a new column.

### **10. Concatenating DataFrames with different columns**

```python
import pandas as pd

df1 = pd.DataFrame({
    "Customer": ["Alice", "Bob"],
    "Age": [25, 30],
    "City": ["Delhi", "Mumbai"]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Customer": ["Charlie", "David"],
    "Salary": [50000, 60000],
    "City": ["Chennai", "Bangalore"]
})

print(df2)
```

```python
result = pd.concat([df1, df2])

print(result)
```

Pandas performed three operations:
1. Identified all unique column names from both DataFrames.
2. Created a combined DataFrame containing all those columns.
3. Filled unavailable column values with NaN.
